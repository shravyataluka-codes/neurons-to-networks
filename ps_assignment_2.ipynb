```python
# Cell 1: Import libraries
import pandas as pd
import numpy as np

# Cell 2: Load dataset
df = pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.csv")

# Cell 3: Inspect dataset
print(df.head())
print(df.info())

# Cell 4: Drop redundant columns
df = df.drop(columns=["EmployeeCount", "StandardHours", "Over18", "EmployeeNumber"])
```

```python
# Cell 5: Attrition distribution
attrition_counts = df['Attrition'].value_counts()
print(attrition_counts)

# Cell 6: Percentage
print("Attrition Ratio:", attrition_counts['Yes'] / attrition_counts.sum())
```

```python
# Cell 7: Encode categorical variables
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE

# Encode categorical features
df_encoded = pd.get_dummies(df, drop_first=True)

# Separate features and target
X = df_encoded.drop('Attrition_Yes', axis=1)
y = df_encoded['Attrition_Yes']

# Stratified split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# Apply SMOTE on training set
smote = SMOTE(random_state=42)
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)
```

```python
# Cell 8: Train models
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, roc_auc_score

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier()
}

for name, model in models.items():
    model.fit(X_train_res, y_train_res)
    y_pred = model.predict(X_test)
    print(f"\n{name} Results:")
    print(classification_report(y_test, y_pred))
    print("ROC-AUC:", roc_auc_score(y_test, model.predict_proba(X_test)[:,1]))
```

```python
# Cell 9: Predict Overtime
X_ot = df_encoded.drop('OverTime_Yes', axis=1)
y_ot = df_encoded['OverTime_Yes']

X_train_ot, X_test_ot, y_train_ot, y_test_ot = train_test_split(
    X_ot, y_ot, test_size=0.2, stratify=y_ot, random_state=42
)

for name, model in models.items():
    model.fit(X_train_ot, y_train_ot)
    y_pred_ot = model.predict(X_test_ot)
    print(f"\n{name} Overtime Results:")
    print(classification_report(y_test_ot, y_pred_ot))
```

```python
# Cell 10: Predict JobRole
y_role = df['JobRole']
X_role = df_encoded.drop(columns=[col for col in df_encoded.columns if "JobRole_" in col])

X_train_role, X_test_role, y_train_role, y_test_role = train_test_split(
    X_role, y_role, test_size=0.2, stratify=y_role, random_state=42
)

for name, model in models.items():
    model.fit(X_train_role, y_train_role)
    y_pred_role = model.predict(X_test_role)
    print(f"\n{name} Job Role Results:")
    print(classification_report(y_test_role, y_pred_role))
```

```python
# Cell 11: Feature importance from Random Forest
rf = RandomForestClassifier()
rf.fit(X_train_res, y_train_res)

importances = pd.Series(rf.feature_importances_, index=X_train_res.columns)
top_features = importances.sort_values(ascending=False).head(10)
print("Top Features Driving Attrition:\n", top_features)
```
